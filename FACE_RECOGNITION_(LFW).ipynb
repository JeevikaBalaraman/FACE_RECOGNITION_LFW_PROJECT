```python
import os

dataset_path = "/content/drive/MyDrive/Labelled Faces in the Wild/archive (6)/lfw-deepfunneled/lfw-deepfunneled"

print(os.listdir(dataset_path)[:5])
```

```python
import cv2
import numpy as np
from tqdm import tqdm

from keras_facenet import FaceNet

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
embedder = FaceNet()

print("FaceNet Loaded Successfully")
```

```python
X = []
y = []

image_size = (160,160)

people_folders = os.listdir(dataset_path)

selected_people = []

for person in people_folders:

    person_folder = os.path.join(dataset_path, person)

    if not os.path.isdir(person_folder):
        continue

    image_count = len(os.listdir(person_folder))

    if image_count >= 20:
        selected_people.append(person)

selected_people = selected_people[:5]

print("Selected Persons:", selected_people)

for person in tqdm(selected_people):

    person_folder = os.path.join(dataset_path, person)

    image_files = os.listdir(person_folder)

    for image_name in image_files[:20]:

        image_path = os.path.join(person_folder, image_name)

        try:

            img = cv2.imread(image_path)

            if img is None:
                continue

            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

            img = cv2.resize(img, image_size)

            img = np.expand_dims(img, axis=0)

            embedding = embedder.embeddings(img)

            X.append(embedding[0])
            y.append(person)

        except:
            pass

print("Embeddings Created Successfully")
```

```python
X = np.asarray(X)
y = np.asarray(y)

print(X.shape)
print(y.shape)
```

```python
encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42
)
```

```python
model = SVC(kernel='rbf')

model.fit(X_train, y_train)

print("Training Completed")
```

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy * 100)
```
